## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
#Convert SSN to string type to avoid data loss
df['SSN']= df['SSN'].astype(str)

#Drop unnecessary columns
df = df.drop(columns=['ID', 'Name', 'SSN'])

#Convert counts/days/age to integer
int_cols = [
    'Age', 'Num_Bank_Accounts', 'Num_Credit_Card', 'Num_of_Loan',
    'Num_of_Delayed_Payment', 'Num_Credit_Inquiries', 'Credit_History_Age',
    'Delay_from_due_date', 'Month'
]

df[int_cols] = df[int_cols].astype(int)

#Round financial and ratio columns
float_round_2 = [
    'Annual_Income', 'Monthly_Inhand_Salary', 'Outstanding_Debt',
    'Total_EMI_per_month', 'Amount_invested_monthly', 'Monthly_Balance',
    'Changed_Credit_Limit', 'Interest_Rate'
]
df[float_round_2] = df[float_round_2].round(2)
# For ratios, 4 decimals
df['Credit_Utilization_Ratio'] = df['Credit_Utilization_Ratio'].round(4)
# Fill numeric columns with median
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

# Fill categorical columns with mode
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
````

### What all manipulations have you done and insights you found?

✅ Manipulations Performed

1.   Converted SSN to string type

     *   Reason: To avoid data loss and ensure unique identifiers are preserved.

2.   Dropped unnecessary columns (ID, Name, SSN)
     *   Reason: These columns don’t add analytical value and would only create noise.
     
3.   Converted count/day/age related columns to integer

     *   Columns: Age, Num_Bank_Accounts, Num_Credit_Card, Num_of_Loan, Num_of_Delayed_Payment, Num_Credit_Inquiries, Credit_History_Age, Delay_from_due_date, Month.

     *   Reason: These represent whole numbers, so integers are the appropriate data type.

4.   Rounded financial and ratio-related columns

     *   Rounded financial values (e.g., Annual_Income, Outstanding_Debt, Monthly_Balance) to 2 decimals for consistency.

     *   Rounded Credit_Utilization_Ratio to 4 decimals for precision.

5.   Handled missing values in numerical columns

     *   Strategy: Filled missing values with the median of each column.

     *   Reason: Median is robust to outliers and ensures data distribution remains realistic.

6.   Handled missing values in categorical columns

     *   Strategy: Filled missing values with the mode (most frequent category).

     *   Reason: Maintains categorical distribution and avoids introducing bias.Answer Here.
